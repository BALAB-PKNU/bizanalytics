## ▶ 구간 확인 · 두 배수는 얼마나 흔들리나

교재 4-2 5절처럼 같은 날의 시간들을 한 묶음으로 보고, 과제 3·4에서 쓴 식 그대로 95% 신뢰구간을 계산해 배수로 바꿉니다. 계산 코드는 실행만 합니다.

```python
# ▶ 그대로 실행하세요 — 과제 3·4 배수의 날짜로 묶은 95% 신뢰구간 (교재 4-2 5절과 같은 계산)
if reg_model is not None and cas_model is not None:
    days = h["dteday"].factorize()[0]
    intervals = {}
    for name, model in [("registered", reg_model), ("casual", cas_model)]:
        refit = smf.ols(model.model.formula, data=h).fit(cov_type="cluster", cov_kwds={"groups": days})
        intervals[name] = np.exp(refit.conf_int().loc["rain"]).values
    display(pd.DataFrame(intervals, index=["하한", "상한"]).round(3))
else:
    print("과제 3·4를 먼저 작성하세요.")
```

💬 **결과 읽기** — 회원과 비회원의 구간을 나란히 보세요. 두 구간이 겹치는지만으로 두 배수의 차이를 검정했다고 말하지는 않습니다. 과제 7에서 차이를 얼마나 믿을 수 있는지 판단할 때 씁니다.
